13. Write a Python program to generate all permutations of a given string.

```python
def find_permutation(string):
  if len(string) == 1:
    return [string]
  else:
    perms = []
    for i in range(len(string)):
      for perm in find_permutation(string[:i] + string[i+1:]):
        perms.append(string[i] + perm)
    return perms
```

```python
find_permutation("abc")
```

14. Implement a function to calculate the Fibonacci sequence up to a given number of terms.

```python
def find_fibonaci(num):
  x,y = 0,1
  fib = [0,1]

  for i in range(2,num):
    next = x + y
    fib.append(next)
    x,y = y,next
  return fib
```

```python
find_fibonaci(20)
```

18. Implement a function to find the maximum subarray sum in a given list.

```python
def find_max_sub_array_sum(my_list):
  max_sum = my_list[0]
  current_sum = my_list[0]
  for i in range(1,len(my_list)):
    current_sum = max(my_list[i], current_sum + my_list[i])
    max_sum = max(max_sum, current_sum)
  return max_sum
```

```python
arr = [2, 3, -8, 7, -1, 2, 3]
```

```python
find_max_sub_array_sum(arr)
```

23. Write a program to find the prime factors of a given number.

```python
def find_prime_factor(num):
    prime_factors = []
    divisor = 2

    while divisor <= num:
        if num % divisor == 0:
            prime_factors.append(divisor)
            num //= divisor
        else:
            divisor += 1

    return prime_factors
```

```python
num = 56
find_prime_factor(num)
```

26. Implement a function to find the mode of a list of numbers.

```python
def find_mode(my_list):
  count = {}
  for ele in my_list:
    if ele in count:
      count[ele] += 1
    else:
      count[ele] = 1
      max_count = max(count.values())
  return [key for key,value in count.items() if value == max_count]
```

```python
find_mode([1, 2, 2, 3, 4, 4, 5, 6, 7, 7, 10, 10, 25])
```

```python
find_mode([1,2,3,3,3,3,5,5,5])
```

34. Implement a function to find the longest common prefix among a list of strings.

```python
def find_longest_common_prefix(strings):
  if len(strings)==0:
    return ""
  smalest_string = min(strings,key=len)
  longest_common_prefix =""
  for char in smalest_string:
    for i,string in enumerate(strings):
      if string[i] == char:
        longest_common_prefix += char
  return longest_common_prefix

strings = ["flower","flow","flight"]
print(find_longest_common_prefix(strings))
```

42. Implement a function to find the longest palindrome substring in a given string.

```python
def find_longest_palindrome_substring(string):
  longest_palindrome = ""
  for i in range(len(string)):
    for j in range(i,len(string)):
      if string[i:j+1] == string[i:j+1][::-1] and len(string[i:j+1]) > len(longest_palindrome):
        longest_palindrome = string[i:j+1]
  return longest_palindrome
```

44. Implement a function to check if a given number is a perfect number.

```python
def find_perfect_num(num):
  if num <= 0:
    return False
  divisors = [i for i in range(1,num) if num % i == 0]
  return sum(divisors) == num
```
